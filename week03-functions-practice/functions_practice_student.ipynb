### Concept Recap (quick — no slides needed today)

- **Composition** — a function's body can call another function you've already defined.
- **Parameter vs. argument** — the name in `def` vs. the value you pass in when calling.
- **Default parameters** — `def f(x, y=10):` lets a caller skip `y`.
- **`return` vs. `print`** — `print` only displays a value; `return` hands it back so it can be used, stored, or passed into another function.
- A function that never hits a `return` (or only calls `print`) hands back `None` — that's often the first clue something's wrong.

### Debug 5 — `average_of_two`

Runs with no error — but Python doesn't compute left to right here. Work out by hand what order the operations in `return a + b / 2` actually happen in.

```python
def average_of_two(a, b):
    return a + b / 2

print(average_of_two(10, 20))
```

### Debug 6 — `bmi_report`

**Expected:** `bmi_report(1.75, 70)` should return the same value as `bmi(70, 1.75)` — about `22.86`.

`bmi_report` calls `bmi` for you, so it runs and prints *something* with no error. But arguments are matched to parameters by **position**, not by what you call them in your head — check the order `bmi_report` passes its arguments into `bmi` against the order `bmi` expects them.

```python
def bmi(weight_kg, height_m):
    return weight_kg / height_m ** 2

def bmi_report(height_m, weight_kg):
    return bmi(height_m, weight_kg)

print(bmi_report(1.75, 70))
```

### Debug 7 — `apply_discount`

**Expected:** `apply_discount(10)` should return `900.0` — the global `total_price` of `1000` with a `10%` discount taken off.

This one raises `UnboundLocalError`, which looks bizarre because `total_price` clearly exists (look at the line right above the function). The rule: if a function *assigns* to a name anywhere in its body, Python treats that name as local for the **entire** function — even on lines before the assignment. Fix it by passing `total_price` in as a parameter instead of reaching out for the global variable.

```python
total_price = 1000

def apply_discount(discount_percent):
    total_price = total_price - total_price * discount_percent / 100
    return total_price

print(apply_discount(10))
```

### Debug 8 — `hypotenuse`

**Expected:** `hypotenuse(3, 4)` should return `5.0`.

The error message names a function and complains about how many arguments it "takes" versus how many "were given." `hypotenuse` is trying to square both legs in a single call to `square` — but `square` only has room for one number at a time. Fix the call so `square` is called twice and the two results are added, *then* raised to the power `0.5`.

```python
def square(x):
    return x ** 2

def hypotenuse(leg_a, leg_b):
    return square(leg_a, leg_b) ** 0.5

print(hypotenuse(3, 4))
```
