### Concept Recap (quick — no slides needed today)

- **Composition** — a function's body can call another function you've already defined.
- **Parameter vs. argument** — the name in `def` vs. the value you pass in when calling.
- **Default parameters** — `def f(x, y=10):` lets a caller skip `y`.
- **`return` vs. `print`** — `print` only displays a value; `return` hands it back so it can be used, stored, or passed into another function.
- A function that never hits a `return` (or only calls `print`) hands back `None` — that's often the first clue something's wrong.
